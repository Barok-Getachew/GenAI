```python
import chromadb

client = chromadb.Client()

```

```python
collection=client.create_collection("news_articles")
```

```python
collection.add(
    ids=["id1", "id2", "id3", "id4"],
    documents=[
        "Biruk is a computer science student at the University of Salzburg.",
        "Flutter is a framework for building cross-platform mobile apps.",
        "ChromaDB is a vector database for embeddings and semantic search.",
         "Tesla is a number electric car producer."
    ],
    metadatas=[
        {"category": "person"},
        {"category": "technology"},
        {"category": "database"}, {"category": "database"}
    ]
)
```

```python
data = collection.get(include=["documents","embeddings","metadatas"])


for i , emb in enumerate(data["embeddings"]):
    print(f"\nDocument {i+1}")
    print("Text:", data["documents"][i])
    print("Embedding length:",len(emb))
    print("First 10 values:", emb[:10])
```

```python
results = collection.query(
    query_texts = ["This is a query related to elon musk"],
    n_results = 2
)
print(results)
```
